```python
import pandas as pd
from mlxtend.frequent_patterns import fpgrowth
from mlxtend.preprocessing import TransactionEncoder

transakcje = [
    ["smartfon", "etui", "szkło hartowane"],
    ["smartfon", "etui", "szkło hartowane", "ładowarka"],
    ["smartfon", "słuchawki", "powerbank"],
    ["smartfon", "etui", "powerbank"],
    ["smartfon", "etui", "szkło hartowane", "słuchawki"],
    ["smartfon", "ładowarka", "powerbank"],
    ["smartwatch", "słuchawki", "ładowarka"],
    ["etui", "szkło hartowane"],
    ["smartfon", "smartwatch", "słuchawki"],
    ["powerbank", "ładowarka"],
]

N = len(transakcje)


def lt(nazwy_przedmiotów: list[str]):
    przedmioty = set(nazwy_przedmiotów)
    return sum(1 for t in transakcje if all(p in t for p in przedmioty))


lt_etui = lt(["etui"])
lt_etui_szkło = lt(["etui", "szkło hartowane"])
lt_szkło = lt(["szkło hartowane"])

confidence_etui_szkło = lt_etui_szkło / lt_etui
confidence_szkło_etui = lt_etui_szkło / lt_szkło

print("lt_etui:", lt_etui)
print("confidence {etui} -> {szkło hartowane}:", confidence_etui_szkło)
print("confidence {szkło hartowane} -> {etui}:", confidence_szkło_etui)
```

```python
lt_etui = lt(["etui"])
lt_szkło = lt(["szkło hartowane"])
lt_etui_szkło = lt(["etui", "szkło hartowane"])

support_etui = lt_etui / N
support_szkło = lt_szkło / N

confidence_etui_szkło = lt_etui_szkło / lt_etui
confidence_szkło_etui = lt_etui_szkło / lt_szkło

lift_etui_szkło = confidence_etui_szkło / support_szkło
lift_szkło_etui = confidence_szkło_etui / support_etui

print("lift {etui} -> {szkło hartowane}:", lift_etui_szkło)
print("lift {szkło hartowane} -> {etui}:", lift_szkło_etui)
```

```python
encoder = TransactionEncoder()
macierz_transakcji = encoder.fit_transform(transakcje)
df_transakcje = pd.DataFrame(macierz_transakcji, columns=encoder.columns_)  # type: ignore

częste_zbiory_fpgrowth = fpgrowth(df_transakcje, min_support=0.3, use_colnames=True)

print(częste_zbiory_fpgrowth)
```
