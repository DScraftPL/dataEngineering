```python
import pandas as pd

sprzedaz_09_10 = pd.read_csv('data/online_retail_09_10.csv', sep=',')
sprzedaz_10_11 = pd.read_csv('data/online_retail_10_11.csv', sep=',')

print(sprzedaz_09_10.columns.values)
print(sprzedaz_10_11.columns.values)

print(all(sprzedaz_09_10.columns == sprzedaz_10_11.columns))

sprzedaz = pd.concat([sprzedaz_09_10, sprzedaz_10_11], ignore_index=True)
```

```python
print(sprzedaz_09_10.shape) #wynik: (525461, 8)
print(sprzedaz_10_11.shape) #wynik: (541910, 8)
print(sprzedaz.shape) #wynik: (1067371, 8)
print(sprzedaz_09_10.shape[0] + sprzedaz_10_11.shape[0]
      == sprzedaz.shape[0]) #wynik: true
```

```python
f1 = sprzedaz.loc[sprzedaz['InvoiceNo'] == 'C489449', :]
f2 = sprzedaz.loc[sprzedaz['InvoiceNo'] == '489450', :]

print(f1[["Quantity", "InvoiceDate"]].values)
print(f2[["Quantity", "InvoiceDate"]].values)

print(sprzedaz.loc[sprzedaz['CustomerID'] == 16321.0, ['InvoiceNo']].drop_duplicates())
```

```python
zwroty = sprzedaz['InvoiceNo'].str.startswith('C')
francja = sprzedaz['Country'] == 'France'

print(sprzedaz.loc[zwroty & francja, ['InvoiceNo','CustomerID']])
klient_12615 = (sprzedaz['CustomerID'] == 12615.0)

print(sprzedaz.loc[zwroty & klient_12615, ['Description', 'StockCode']])
print(sprzedaz[klient_12615 & (sprzedaz['StockCode'] == '22618')])
```

```python
zwroty = sprzedaz['InvoiceNo'].str.startswith('C')
print(sum(zwroty)) #wynik: 19494
print(sum(zwroty) / len(sprzedaz)) #wynik: 0.018
print(sprzedaz.loc[zwroty, "InvoiceNo"].nunique()) #wynik: 8292

print((sprzedaz.loc[zwroty, 'Quantity'] < 0).all()) #wynik: False
print(sprzedaz.loc[zwroty & (sprzedaz["Quantity"] > 0)])
```

```python
neg = sprzedaz.loc[
    (sprzedaz["Quantity"] < 0) & (~sprzedaz["InvoiceNo"].str.startswith("C"))
]
print(len(neg))
print(neg.head())

print((neg['UnitPrice'] == 0).all())
print((neg['CustomerID'].isna()).all())
```
