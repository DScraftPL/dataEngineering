```python
df = pd.read_csv("data/auto-mpg.csv")

df.info()
```

```python
# korelacja na danych liczbowych, pozbywamy się zbędnych danych

df = df.drop(columns=["car name"])
df.info()
```

```python
df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce") # jeżeli pojawi się błąd, która nie może być rzutowana, zastępujemy ją pustą komurką
df.info() # jest 6 nulli
```

```python
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].median())
df.info()
```

```python
import seaborn as sns

corr_mat = df.corr()
sns.heatmap(corr_mat, annot=True, cmap="coolwarm")
```

```python
import yfinance as yf

df = yf.Ticker("AAPL").history(
    start="2020-01-01",
    end="2022-12-31",
    interval="1d",
    actions=False
)

df.info()
```

```python
df = df[["Close"]].copy()
df.head()
```

```python
import matplotlib.pyplot as plt

df.plot()
plt.grid()
plt.show()
```

```python
print(df['Close'].autocorr(lag=1)) #opóźnienie

# widać, że jest silna autokorelacja. 

for lag in [1,2,20,250]:
    print(df['Close'].autocorr(lag=lag))
```

# ZADANIE 1

```python
df = pd.read_csv("data/melbourne.csv")
df.info()
```

```python
df.head()
```

```python
print(df.isna().mean().sort_values())
```

```python
df = df.drop(columns=["BuildingArea", "YearBuilt"])

df.info()
```

```python
df["Landsize"] = df["Landsize"].fillna(df["Landsize"].mean())
df["Car"] = df["Car"].fillna(df["Car"].median())
df["Bathroom"] = df["Bathroom"].fillna(df["Bathroom"].median())
df["Bedroom"] = df["Bedroom"].fillna(df["Bedroom"].median())
df["Price"] = df["Price"].fillna(df["Price"].median())
df["Latitude"] = df["Latitude"].fillna(df["Latitude"].mean())
df["Longtitude"] = df["Longtitude"].fillna(df["Longtitude"].mean())
df = df.dropna(subset=["Distance", "Postcode", "CouncilArea", "Propertycount"])

df.isna().mean().sort_values()

df.info()
```

```python
df = df.drop(columns = ["Regionname", "CouncilArea", "Date", "SellerG", "Method", "Type", "Address", "Suburb", "ParkingArea"])
corr_mat = df.corr()
sns.heatmap(corr_mat, annot=True, cmap="coolwarm")
```

```python
import yfinance as yf

df = yf.Ticker("AAPL").history(
    start="2020-01-01",
    end="2024-12-31",
    interval="1d",
    actions=False
)

df.info()
```

```python
df_close = df[["Close"]].copy()
df_close.head()
```

```python
for lag in [1,2,30,365]:
    print(df_close['Close'].autocorr(lag=lag))
```

```python
df_open = df[["Open"]].copy()
df_open.head()
```

```python
for lag in [1,2,30,365]:
    print(df_open['Open'].autocorr(lag=lag))
```

```python
df_open["StopaZwrotu"] = df_open.pct_change()

df_open.info()
```

```python
for lag in [1,2,30,365]:
    print(df_open['StopaZwrotu'].autocorr(lag=lag))
```

```python
df_open["KwadratStopaZwrotu"] = df_open["StopaZwrotu"]**2

df_open.head()
```

```python
for lag in [1,2,30,365]:
    print(df_open['KwadratStopaZwrotu'].autocorr(lag=lag))
```
